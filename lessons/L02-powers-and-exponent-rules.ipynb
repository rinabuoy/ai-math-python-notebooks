## 1. The rules
$a^m a^n=a^{m+n}$, $(a^m)^n=a^{mn}$, $(ab)^n=a^nb^n$, $(a/b)^n=a^n/b^n$, $a^m/a^n=a^{m-n}$.

```python
a, b, m, n = 3.0, 2.0, 5, 3
checks = {
 "a^m a^n = a^(m+n)":   np.isclose(a**m * a**n, a**(m+n)),
 "(a^m)^n = a^(mn)":    np.isclose((a**m)**n, a**(m*n)),
 "(ab)^n = a^n b^n":    np.isclose((a*b)**n, a**n * b**n),
 "(a/b)^n = a^n / b^n": np.isclose((a/b)**n, a**n / b**n),
 "a^m / a^n = a^(m-n)": np.isclose(a**m / a**n, a**(m-n)),
}
for k, v in checks.items(): print(f"{k:24s}", v)
```

## 2. Why it matters in ML
- Derivative power rule: $(x^n)'=nx^{n-1}$.
- $e^{-z}$ appears in the sigmoid.
- Logs turn multiplication of powers into addition: $\log(a^m a^n)=(m+n)\log a$.

```python
x = np.linspace(-4, 4, 9)
sigmoid = 1/(1+np.exp(-x))
print(np.round(sigmoid, 3))
print(np.isclose(np.log(2**5 * 2**3), 8*np.log(2)))
```

## 3. Powers grow fast – plot

```python
x = np.linspace(0, 4, 100)
for base in (1.5, 2, np.e):
    plt.plot(x, base**x, label=f"{base:.2f}^x")
plt.legend(); plt.title("Exponential growth"); plt.show()
```
