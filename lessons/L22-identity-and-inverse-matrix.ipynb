## 1. Formula $A^{-1}=\frac1{ad-bc}\begin{pmatrix}d&-b\\-c&a\end{pmatrix}$

```python
def inv2(M):
    (a,b),(c,d)=M; D=a*d-b*c
    if D==0: raise ValueError("no inverse (det=0)")
    return np.array([[d,-b],[-c,a]])/D
A=np.array([[3,4],[1,2]]); print(inv2(A), np.allclose(A@inv2(A),np.eye(2)))   # (1/2)[[2,-4],[-1,3]]
```

## 2. Solving matrix equations – left vs right

```python
M=np.array([[2,-3],[4,3]]); N=np.array([[-4,14],[10,10]])
X=np.linalg.inv(M)@N      # MX = N
Y=N@np.linalg.inv(M)      # YM = N
print(np.round(X,3)); print(np.allclose(M@X,N), np.allclose(Y@M,N), np.allclose(X,Y))
```

## 3. (AB)^-1 = B^-1 A^-1

```python
A=np.array([[2,1],[-5,-4]]); B=np.array([[3,-1],[-1,1]])
print(np.allclose(np.linalg.inv(A@B),np.linalg.inv(B)@np.linalg.inv(A)))
try: inv2([[1,1],[2,2]])
except ValueError as e: print(e)
```
