## 1. Quiz-style examples

```python
from scipy import stats
z99=stats.norm.ppf(0.5+0.495); print(z99, z99*1/np.sqrt(100))      # area 0.495 between 0 and z -> 99% CI, margin 0.258
t=stats.t.ppf(0.975,19); lo,hi=50-t*10/np.sqrt(20),50+t*10/np.sqrt(20); print(t,lo,hi)
assert abs(lo-45.32)<0.01 and abs(hi-54.68)<0.01
```

## 2. Coverage simulation: does a 95% CI cover the truth 95% of the time?

```python
rng=np.random.default_rng(0); mu,sigma,n=10,3,15; hit=0
for _ in range(5000):
    x=rng.normal(mu,sigma,n); m=x.mean(); s=x.std(ddof=1); tt=stats.t.ppf(0.975,n-1)
    hit+= (m-tt*s/np.sqrt(n)<=mu<=m+tt*s/np.sqrt(n))
print(hit/5000)
```

## 3. Why t, not z, for small n

```python
zz=stats.norm.ppf(0.975); hit=0
for _ in range(5000):
    x=rng.normal(mu,sigma,n); m=x.mean(); s=x.std(ddof=1); hit+=(m-zz*s/np.sqrt(n)<=mu<=m+zz*s/np.sqrt(n))
print(hit/5000)   # below 0.95: z with estimated s undercovers
```
