## 1. z-test p-values (sigma known)

```python
from scipy import stats
mu0,sig,n=66.7,3,10; se=sig/np.sqrt(n)
z=(68.442-mu0)/se; print(z,'right',stats.norm.sf(z),'two',2*stats.norm.sf(abs(z)),'left (64.252)',stats.norm.cdf((64.252-mu0)/se))
```

## 2. Critical values and power

```python
k05=mu0+stats.norm.ppf(.95)*se; k01=mu0+stats.norm.ppf(.99)*se; print(k05,k01)
beta=stats.norm.cdf((k05-70)/se); print('beta',beta,'power',1-beta)
for a in (.01,.05,.1): k=mu0+stats.norm.ppf(1-a)*se; print(a,'power at mu=70:',1-stats.norm.cdf((k-70)/se))
```

## 3. Binomial coin tests

```python
print(stats.binom.sf(7,10,.5),stats.binom.sf(79,100,.5))   # 0.0547, 5.6e-10
```

## 4. Simulation: alpha is the false-positive rate; power grows with n

```python
rng=np.random.default_rng(0)
def rej(mu,n,reps=4000): return np.mean([ (rng.normal(mu,3,n).mean()-mu0)/(3/np.sqrt(n))>1.645 for _ in range(reps)])
print(rej(66.7,10),rej(68,10),rej(68,40))
```
