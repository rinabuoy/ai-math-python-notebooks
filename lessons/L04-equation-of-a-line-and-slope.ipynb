## 1. Slope (មេគុណប្រាប់ទិស)

```python
def slope(p, q):
    (x1,y1),(x2,y2)=p,q
    return (y2-y1)/(x2-x1)
print(slope((0,1),(1,3)))   # y = 2x + 1 -> 2
print(slope((0,1),(1,-2)))  # y = -3x + 1 -> -3
```

## 2. Line through two points

```python
def line_through(p, q):
    a = slope(p, q); b = p[1] - a*p[0]
    return a, b
a, b = line_through((2,48.63),(5,48.97))   # group means from Grade 12 basic
print(f"y = {a:.2f}x + {b:.2f}", "-> year 8:", round(a*8+b,2))
```

## 3. Parallel lines never meet (no solution)

```python
A = np.array([[1,3],[1,3]], float); rhs = np.array([6,-3])
print("same slope:", -A[0,0]/A[0,1], -A[1,0]/A[1,1])
print("det =", np.linalg.det(A))   # 0 -> singular, no unique solution
```
