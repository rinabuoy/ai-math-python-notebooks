## 1. Image of a point

```python
A=np.array([[-0.5,1],[4,3]]); print(A@np.array([2,0.5]))   # (-1/2, 19/2) – Grade 11 advanced
```

## 2. Find the matrix from two point→image pairs (textbook method)

```python
P=np.array([[1,-1],[2,1]],float)      # columns: input points (1,2) and (-1,1)
Q=np.array([[8,1],[1,2]],float)       # columns: their images (8,1) and (1,2)
A=Q@np.linalg.inv(P); print(A)         # [[2,3],[-1,1]]
print(A@[1,2], A@[-1,1])
```

## 3. Composition: first A then B is BA

```python
A=np.array([[2,-1],[1,0]]); B=np.array([[-0.5,2],[4,-5]])
print(B@A, (B@A)@[2,-1])   # (3/2, 10)
```

## 4. Fixed points and the inverse

```python
A=np.array([[2,3],[-1,1]],float)
print("inverse:",np.linalg.inv(A)); r=np.linalg.matrix_rank(A-np.eye(2)); print("rank(A-I) =",r, "-> only the origin is fixed" if r==2 else "-> a line of fixed points")
sq=np.array([[0,1,1,0,0],[0,0,1,1,0]]); im=A@sq
plt.plot(*sq,label='square'); plt.plot(*im,label='image'); plt.legend(); plt.axis('equal'); plt.show()
```
