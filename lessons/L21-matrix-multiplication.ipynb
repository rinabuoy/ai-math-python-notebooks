## 1. Row × column

```python
A=np.array([[2,-1],[0,2]]); B=np.array([[3,1],[1,2]])
print(A@B)          # [[5,0],[2,4]]
print(B@A)          # different: AB != BA
```

## 2. Implement from the definition

```python
def matmul(A,B):
    A,B=np.array(A),np.array(B)
    assert A.shape[1]==B.shape[0], "inner dimensions must match"
    return np.array([[sum(A[i,k]*B[k,j] for k in range(A.shape[1])) for j in range(B.shape[1])] for i in range(A.shape[0])])
print(np.array_equal(matmul(A,B),A@B))
```

## 3. Business example from the Cambodian book

```python
buy=np.array([[18,12,11],[16,20,10]]); price=np.array([[7,6],[4,5],[2,2]])
print(buy@price)      # total cost per buyer per shop
trucks=np.array([[7,2,4],[5,6,3]]); cost=np.array([[900,600],[700,500],[500,400]]); print(trucks@cost)   # [[9700,6800],[10200,7200]]
```
