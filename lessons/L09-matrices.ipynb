## 1. A matrix is a table

```python
books=np.array([[13,5,7],[8,10,4]])   # shelves x subjects
print(books, "order:", books.shape)
print("entry a_12 =", books[0,1])
```

## 2. Equality, addition, scalar multiple

```python
A=np.array([[30,20,25],[35,25,23]]); B=np.array([[35,28,20],[40,21,30]])
print(A+B)              # production over two months
print(2*A)
print(np.array_equal(A,A.copy()), np.array_equal(A,B))
```

## 3. Rows = examples, columns = features (ML view)

```python
import pandas as pd
df=pd.DataFrame(books,index=["shelf 1","shelf 2"],columns=["maths","physics","chemistry"]); print(df); print(df.values.shape)
```
