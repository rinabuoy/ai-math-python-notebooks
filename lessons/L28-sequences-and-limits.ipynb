```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Convergent sequences

```python
n=np.arange(1,2001)
b=(n+1)/n; a=(-1.0)**(n+1)/n
print(b[-1], a[-1])     # -> 1 and 0
plt.plot(n[:60],b[:60],'o-',label="(n+1)/n"); plt.plot(n[:60],a[:60],'s-',label="(-1)^(n+1)/n"); plt.legend(); plt.show()
```

## 2. Limit laws: (5n+1)/(3n+2) → 5/3

```python
n=10.0**np.arange(1,8); print((5*n+1)/(3*n+2), 5/3)
```

## 3. Divergence

```python
print((np.arange(1,6)**2), 2-np.arange(1,6))
```

## 4. e as a limit of (1+1/n)^n

```python
for n in (1,2,3,12,365,10**6): print(n, (1+1/n)**n)
print(np.e)
```

## 5. Iterates of Newton's method converge too

```python
x=0.05; seq=[x]
for _ in range(6): x=x-(np.exp(x)-1/x)/(np.exp(x)+1/x**2); seq.append(x)
print(np.round(seq,4))   # -> 0.5671
```
