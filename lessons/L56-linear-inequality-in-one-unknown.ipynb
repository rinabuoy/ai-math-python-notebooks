```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Solve with SymPy and check by testing points

```python
m=sp.symbols('m')
sol=sp.solve_univariate_inequality((1-m)/2+(m+5)/6>4-(3*m-7)/5,m); print(sol)    # m > 61/4
f=sp.lambdify(m,(1-m)/2+(m+5)/6-(4-(3*m-7)/5)); print(f(15)>0,f(15.25)>0,f(15.2)>0)   # 15.25 = 61/4 is the boundary
```

## 2. The sign flip

```python
a=sp.symbols('a')
print(sp.solve_univariate_inequality(-3*a>-4,a))      # a < 4/3 (divided by -3, flipped)
for v in (1,4/3,2): print(v,-3*v>-4)
```

## 3. Number line

```python
fig,ax=plt.subplots(figsize=(7,1.2)); ax.axhline(0,color='k')
ax.plot([15,20],[0,0],lw=6); ax.plot([15],[0],'ko')   # closed dot: 15 is included (>=)
ax.set_xlim(10,20); ax.set_yticks([]); ax.set_title('m >= 15'); plt.show()
```

## 4. From one variable to a half-plane (ML link)

```python
xx,yy=np.meshgrid(np.linspace(-3,3,200),np.linspace(-3,3,200))
plt.contourf(xx,yy,(2*xx+yy-1>0).astype(float),levels=[-.5,.5,1.5],alpha=.4); plt.axline((0,1),slope=-2,color='k'); plt.title('2x+y-1>0'); plt.show()
relu=lambda z: np.maximum(0,z); print(relu(np.array([-2,0,3])))
```

```python
# Solution 1
x=sp.symbols('x'); s1=sp.solve_univariate_inequality(3*(x-2)>=2*x+5,x); s2=sp.solve_univariate_inequality(2*x<2*(x+1),x); print(s1,s2); assert all(bool(3*(v-2)>=2*v+5)==(v>=11) for v in range(0,25)); assert all(2*v<2*(v+1) for v in range(-50,50))   # s2 holds for every x
```

```python
# Solution 2
a=sp.symbols('a'); assert sp.solve_univariate_inequality(a/2+1>2*a-1,a)==sp.solve_univariate_inequality(a+1>4*a-3,a); print('equivalent: a < 4/3')
```
