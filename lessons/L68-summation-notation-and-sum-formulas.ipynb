```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Closed forms with SymPy

```python
n,k=sp.symbols('n k',integer=True,positive=True)
print(sp.factor(sp.summation(k,(k,1,n))),sp.factor(sp.summation(k**2,(k,1,n))),sp.factor(sp.summation(k**3,(k,1,n))))
assert sp.simplify(sp.summation(k**3,(k,1,n))-(n*(n+1)/2)**2)==0
```

## 2. Textbook examples

```python
print(sum(4*i+3 for i in range(1,16)),sum((i+3)*i for i in range(1,21)),sum(i**3 for i in range(12,51)))   # 525 3500 1621269
print(sp.simplify(sp.summation(1/(k*(k+1)),(k,1,n))))                # n/(n+1)
print(sp.simplify(n**3+2*n-((n-1)**3+2*(n-1))))                       # a_n = S_n - S_{n-1}
```

## 3. Σ properties: expanding a squared error

```python
rng=np.random.default_rng(0); a=rng.normal(size=50); b=rng.normal(size=50)
print(np.isclose(((a+b)**2).sum(),(a**2).sum()+2*(a*b).sum()+(b**2).sum()), np.isclose((a-a.mean()).sum(),0))
print('sum(ab) vs sum(a)sum(b):',(a*b).sum(),a.sum()*b.sum())
```

## 4. Σ in ML: mean-squared error and its gradient

```python
x=rng.normal(size=100); y=3*x+1+rng.normal(0,.1,100); w,b0=2.,0.
mse=lambda w,b0: np.mean((y-(w*x+b0))**2)
g_w=-2*np.mean(x*(y-(w*x+b0)))             # d/dw of (1/n) sum (y-wx-b)^2
eps=1e-6; print(g_w,(mse(w+eps,b0)-mse(w-eps,b0))/(2*eps))
```
