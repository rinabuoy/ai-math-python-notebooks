```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Symbolic

```python
x,y=sp.symbols('x y')
for f in (x**2+y**2, 3*x**2*y**3, sp.sin(x*y)): print(f,"| f_x =",sp.diff(f,x),"| f_y =",sp.diff(f,y))
```

## 2. Numeric partials

```python
f=lambda x,y:x**2+y**2
h=1e-6; x0,y0=2,4
print((f(x0+h,y0)-f(x0-h,y0))/(2*h), (f(x0,y0+h)-f(x0,y0-h))/(2*h))   # 4, 8
```

## 3. Slice picture

```python
xs=np.linspace(-3,3,200); plt.plot(xs,f(xs,4),label='y=4 slice'); plt.plot(xs,f(2,xs),label='x=2 slice'); plt.legend(); plt.show()
```
