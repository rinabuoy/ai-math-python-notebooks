```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Population vs sample

```python
pop=np.array([-2,-1,0,1,2.]); smp=np.array([-1,0,2.])
print(pop.mean(), smp.mean(), smp.var(ddof=1), smp.var())   # 0, 1/3, 7/3, 14/9
assert np.isclose(smp.var(ddof=1),7/3)
```

## 2. Bessel's correction by simulation

```python
rng=np.random.default_rng(1)
v0=[rng.normal(0,2,5).var(ddof=0) for _ in range(20000)]; v1=[rng.normal(0,2,5).var(ddof=1) for _ in range(20000)]
print(np.mean(v0),np.mean(v1),'true 4')
```

## 3. Law of large numbers

```python
x=rng.integers(1,7,100000); run=np.cumsum(x)/np.arange(1,100001)
print(run[[9,99,999,9999,99999]])      # approaches 3.5
plt.plot(run); plt.axhline(3.5,color='r'); plt.xscale('log'); plt.show()
```
