## 1. 68-95-99.7

```python
from scipy import stats
print([round(stats.norm.cdf(k)-stats.norm.cdf(-k),4) for k in (1,2,3)])      # 0.6827 0.9545 0.9973
```

## 2. Standardisation

```python
rng=np.random.default_rng(1); x=rng.normal(66.7,3,100000); z=(x-x.mean())/x.std(); print(z.mean().round(3),z.std().round(3))
```

## 3. Sum of independent Gaussians: T~N(10,2^2), L~N(5,1^2)

```python
T=rng.normal(10,2,200000); L=rng.normal(5,1,200000); R=T+L; print(R.mean(),R.var(), 'theory 15 and', 2**2+1**2)
print(stats.kstest((R-15)/np.sqrt(5),'norm').pvalue>0.01)
```

## 4. Binomial -> Gaussian (correct z-scores!)

```python
n,p=100,.5; mu,sd=n*p,np.sqrt(n*p*(1-p)); print(sd)
print('exact P(X>=80)',stats.binom.sf(79,n,p),'normal approx z=',(80-mu)/sd,stats.norm.sf((79.5-mu)/sd))
n,p=400,.5; mu,sd=n*p,np.sqrt(n*p*(1-p)); print('P(250<=X<=350) ~',stats.binom.cdf(350,n,p)-stats.binom.cdf(249,n,p))
```
