## 1. One-sample t-test from summary statistics

```python
from scipy import stats
n,xb,s=10,68.442,3.113; t=(xb-66.7)/(s/np.sqrt(n)); print(t,stats.t.sf(t,n-1),2*stats.t.sf(t,n-1),'left',stats.t.cdf((64.252-66.7)/(s/np.sqrt(n)),n-1))
```

## 2. Welch two-sample test

```python
r=stats.ttest_ind_from_stats(68.442,3.113,10,65.949,3.106,9,equal_var=False); print(r.statistic,r.pvalue/2,r.pvalue)
sx,sy=3.113**2/10,3.106**2/9; print('df',(sx+sy)**2/(sx**2/9+sy**2/8))
```

## 3. Paired test

```python
d=np.array([0.4,1.3,2.7,1.1,1.2,-1.3,2.8,1.9,2.2,-1.4]); t=d.mean()/(d.std(ddof=1)/np.sqrt(10)); print(d.mean(),d.std(ddof=1),t,stats.t.sf(t,9))
print(stats.ttest_1samp(d,0,alternative='greater'))
```

## 4. A/B tests: mean and conversion

```python
s1,s2=10**2/80,15**2/20; t=-5/np.sqrt(s1+s2); df=(s1+s2)**2/(s1**2/79+s2**2/19); print(t,df,stats.t.cdf(t,df))
x,y,na,nb=20,8,80,20; pp=(x+y)/(na+nb); z=(x/na-y/nb)/np.sqrt(pp*(1-pp)*(1/na+1/nb)); print(pp,z,stats.norm.cdf(z))
```

```python
# Solution 1
rng=np.random.default_rng(0); base=rng.normal(200,30,10); a=base+rng.normal(0,1,10); b=a-1+rng.normal(0,1,10); print(stats.ttest_rel(a,b).pvalue,stats.ttest_ind(a,b,equal_var=False).pvalue); assert stats.ttest_rel(a,b).pvalue<stats.ttest_ind(a,b,equal_var=False).pvalue
```
