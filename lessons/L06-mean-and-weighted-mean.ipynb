## 1. Mean

```python
visitors=np.array([18,12,30,26,38,42,48])
print(visitors.mean())     # 30.5 (Grade 9)
```

## 2. Weighted mean from a frequency table

```python
x=np.array([10,12,14,16,17,18,20]); f=np.array([3,3,4,7,1,5,7])
print((x*f).sum(), f.sum(), (x*f).sum()/f.sum())   # 481, 30, 16.03
print(np.average(x, weights=f))
```

## 3. The mean is the balance point – it minimises Σ(x−a)²

```python
data=np.array([2.,3.,10.])
cands=np.linspace(0,12,1201)
loss=[((data-c)**2).sum() for c in cands]
print("argmin:", cands[np.argmin(loss)], "mean:", data.mean())
plt.plot(cands,loss); plt.axvline(data.mean(),c='r'); plt.title("sum of squared distances"); plt.show()
```
