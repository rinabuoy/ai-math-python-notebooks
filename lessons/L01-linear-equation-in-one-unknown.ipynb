## 1. The balance-scale idea
An equation `a*x + b = c` (with `a != 0`) says two sides weigh the same. Doing the same operation to both sides keeps the balance, so `x = (c - b) / a`.

```python
def solve_linear(a, b, c):
    """Solve a*x + b = c for x."""
    if a == 0:
        raise ValueError("a must be non-zero")
    return (c - b) / a

x = solve_linear(1, 86, 94)      # the textbook equation 86 + x = 94
print("x =", x)
print("check:", 86 + x == 94)
```

## 2. Try values, then check (Grade 7 method)
The book first *tries* values and compares both sides. A loop does the same.

```python
for x in [12, 10, 8]:
    left = 86 + x
    print(f"x={x:>2}: left={left}, right=94 ->", "root" if left == 94 else "not a root")
```

## 3. Symbolic solution with SymPy

```python
x = sp.symbols('x')
print(sp.solve(sp.Eq(3*x + 5, 20), x))   # [5]
print(sp.solve(sp.Eq(x/2 - 4, 3), x))    # [14]
```
