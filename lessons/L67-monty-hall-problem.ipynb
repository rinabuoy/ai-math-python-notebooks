## 1. Simulation

```python
rng=np.random.default_rng(0)
def play(switch,n=3,k=1):
    car=rng.integers(n); choice=rng.integers(n)
    openable=[i for i in range(n) if i not in (car,choice)]
    opened=set(rng.choice(openable,k,replace=False))
    if not switch: return choice==car
    rest=[i for i in range(n) if i!=choice and i not in opened]
    return rng.choice(rest)==car
for sw in (False,True): print('switch' if sw else 'stay',np.mean([play(sw) for _ in range(40000)]))
```

## 2. Exact enumeration (you pick door 0)

```python
from fractions import Fraction
win_switch=Fraction(0)
for car in range(3):
    openable=[d for d in (1,2) if d!=car]
    for o in openable:
        p=Fraction(1,3)/len(openable); other=[d for d in (1,2) if d!=o][0]; win_switch+=p*(other==car)
print(win_switch)   # 2/3
```

## 3. Bayes: P(car at 2 | host opens 3)

```python
prior=[Fraction(1,3)]*3; like=[Fraction(1,2),Fraction(1),Fraction(0)]    # P(opens door 3 | car at door 1,2,3), you chose door 1
post=[p*l for p,l in zip(prior,like)]; Z=sum(post); print([x/Z for x in post])
```

## 4. Generalised game

```python
f=lambda n,k:(n-1)/(n*(n-k-1))
for n,k in ((3,1),(5,1),(5,3),(10,4)):
    sim=np.mean([play(True,n,k) for _ in range(40000)]); print(n,k,round(f(n,k),4),round(sim,3))
```
