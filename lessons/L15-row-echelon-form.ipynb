## 1. SymPy's rref

```python
M=sp.Matrix([[1,2,3],[0,1,4],[0,0,1]]); print(M.rref())
M2=sp.Matrix([[5,1],[10,2]]); print(M2.rref(), "rank", M2.rank())
```

## 2. Own implementation (pivots normalised to 1)

```python
def rref(A,tol=1e-12):
    M=np.array(A,float); r=0; piv=[]
    for c in range(M.shape[1]):
        if r>=M.shape[0]: break
        p=r+np.argmax(abs(M[r:,c]))
        if abs(M[p,c])<tol: continue
        M[[r,p]]=M[[p,r]]; M[r]/=M[r,c]
        for i in range(M.shape[0]):
            if i!=r: M[i]-=M[i,c]*M[r]
        piv.append(c); r+=1
    return M,piv
R,piv=rref([[1,1,1],[1,2,1],[1,1,2]]); print(R,piv)
R,piv=rref([[1,1,1],[2,2,2],[3,3,3]]); print(R,"rank =",len(piv))
```
