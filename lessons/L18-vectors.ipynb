## 1. Sum, difference, scalar multiple

```python
u=np.array([4,1]); v=np.array([1,3])
print(u+v, u-v, 3*np.array([1,2]), -2*np.array([1,2]))
```

## 2. Norms and distances

```python
w=np.array([4,3]); print(np.linalg.norm(w,2), np.linalg.norm(w,1), np.degrees(np.arctan2(3,4)))   # 5, 7, 36.87
a,b=np.array([6,2]),np.array([1,5])
cos=a@b/(np.linalg.norm(a)*np.linalg.norm(b)); print("L1",abs(a-b).sum(),"L2",np.linalg.norm(a-b),"cosine distance",1-cos)
```

```python
fig,ax=plt.subplots()
for vec,c in ((u,'b'),(v,'g'),(u+v,'r')): ax.quiver(0,0,*vec,angles='xy',scale_units='xy',scale=1,color=c)
ax.set_xlim(-1,6); ax.set_ylim(-1,5); ax.grid(True); ax.set_aspect('equal'); plt.show()
```
