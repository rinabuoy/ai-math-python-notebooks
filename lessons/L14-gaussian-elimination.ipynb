## 1. Elementary row operations

```python
def swap(M,i,j): M[[i,j]]=M[[j,i]]
def scale(M,i,k): M[i]*=k
def add_multiple(M,i,j,k): M[j]+=k*M[i]    # row_j += k * row_i
```

## 2. Solve a 3×3 system step by step

```python
Ab=np.array([[1,1,2,12],[3,-3,-1,3],[2,-1,6,24]],float)
M=Ab.copy()
add_multiple(M,0,1,-3); add_multiple(M,0,2,-2)
print(M)
scale(M,1,1/M[1,1]); add_multiple(M,1,2,-M[2,1]); print(M)
z=M[2,3]/M[2,2]; y=M[1,3]-M[1,2]*z; x=M[0,3]-M[0,1]*y-M[0,2]*z
print("solution:",x,y,z)   # 4, 2, 3
```

## 3. General routine with partial pivoting

```python
def gauss_solve(A,b):
    n=len(b); M=np.c_[np.array(A,float),np.array(b,float)]
    for i in range(n):
        p=i+np.argmax(abs(M[i:,i])); swap(M,i,p)
        if abs(M[i,i])<1e-12: raise ValueError("singular system")
        scale(M,i,1/M[i,i])
        for j in range(n):
            if j!=i: add_multiple(M,i,j,-M[j,i])
    return M[:,-1]
print(gauss_solve([[2,5],[8,1]],[46,32]))   # 3, 8 (Coursera quiz)
```

```python
try: gauss_solve([[1,1],[2,2]],[10,24])
except ValueError as e: print("caught:",e)
```

The Grade 9 elimination method (**វិធីបំបាត់**) is exactly step 1–2 above with two unknowns; the book then substitutes back.

```python
print(gauss_solve([[3,1],[2,-3]],[7,12]))   # y=7-3x and 2x-3y=12 -> (3,-2)
```
