## 1. Distance formula

```python
def dist(A, B):
    return float(np.sqrt((B[0]-A[0])**2 + (B[1]-A[1])**2))
print(dist((3,7),(8,-5)))      # 13
print(dist((-1,-2),(3,4)), 2*np.sqrt(13))
```

## 2. Right-triangle test (Grade 9 exercise)

```python
A,B,C=(-2,6),(-1,-1),(1,0)
AB2,AC2,BC2=dist(A,B)**2,dist(A,C)**2,dist(B,C)**2
print(AB2,AC2,BC2, np.isclose(AC2+BC2,AB2))   # right angle at C
```

## 3. It is just the L2 norm of the difference

```python
u,v=np.array([6,2]),np.array([1,5])
print("L2:", np.linalg.norm(u-v), " L1:", np.abs(u-v).sum())   # 5.83 and 8 (Coursera example)
```
