## 1. Cross product

```python
u=np.array([1,-4,1]); v=np.array([2,3,0])
w=np.cross(u,v); print(w, np.linalg.norm(w)**2, "perp:", w@u, w@v)   # (-3,2,11), 134
```

## 2. |u×v| = |u||v| sinθ = parallelogram area

```python
cos=u@v/(np.linalg.norm(u)*np.linalg.norm(v)); sin=np.sqrt(1-cos**2)
print(np.linalg.norm(w), np.linalg.norm(u)*np.linalg.norm(v)*sin)
```

## 3. Scalar triple product = volume = 3×3 determinant

```python
a,b,c=np.array([3,-5,1]),np.array([0,2,-2]),np.array([3,1,1])
print(a@np.cross(b,c), np.linalg.det(np.array([a,b,c])))   # 36
print("tetrahedron:", abs(a@np.cross(b,c))/6)
```

```python
print(np.linalg.det(np.array([[1,0,1],[0,1,1],[1,1,2]])))   # 0 -> coplanar -> dependent
```
