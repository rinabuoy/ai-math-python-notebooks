```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. One house, three power lines

```python
a=np.array([1.,4.,10.])
x=np.linspace(0,12,1201); L=((x[:,None]-a)**2).sum(axis=1)
print("argmin", x[np.argmin(L)], "mean", a.mean())
plt.plot(x,L); plt.axvline(a.mean(),c='r'); plt.title("total cost = total area of squares"); plt.show()
```

## 2. Derivative argument with SymPy

```python
x=sp.symbols('x'); a1,a2,a3=sp.symbols('a1 a2 a3'); L=(x-a1)**2+(x-a2)**2+(x-a3)**2
print(sp.solve(sp.diff(L,x),x))
```

## 3. MSE of a constant predictor

```python
y=np.array([30.,45.,50.,20.])
for c in (y.mean(),y.mean()+5): print(c, np.mean((y-c)**2))
```
