## 1. The children example (age X, height Y)

```python
import pandas as pd
age=[7,7,7,8,8,9,9,9,9,10]; ht=[45,46,46,47,47,49,49,49,50,50]   # textbook-style data: 10 children
J=pd.crosstab(pd.Series(age,name='age'),pd.Series(ht,name='height'))/len(age)
print(J); print('marginal age\n',J.sum(axis=1)); print('marginal height\n',J.sum(axis=0))
print('P(Y=49 | X=9) =',J.loc[9,49]/J.loc[9].sum())        # 3/4
```

## 2. Two dice: independent vs sum

```python
import itertools
pairs=list(itertools.product(range(1,7),repeat=2))
J1=pd.crosstab(pd.Series([a for a,b in pairs],name='X'),pd.Series([b for a,b in pairs],name='Y'))/36
print(np.allclose(J1.values,np.outer(J1.sum(axis=1),J1.sum(axis=0))))        # independent
J2=pd.crosstab(pd.Series([a for a,b in pairs],name='X'),pd.Series([a+b for a,b in pairs],name='sum'))/36
print(J2.loc[3,7]*36, J2.loc[1].get(1,0), J2.sum(axis=0).loc[10]*36)         # 1, 0, 3
print(np.allclose(J2.values,np.outer(J2.sum(axis=1),J2.sum(axis=0))))        # NOT independent
```

## 3. Conditional table = row / row sum

```python
cond=J2.div(J2.sum(axis=1),axis=0); print(cond.loc[4].round(3).to_dict())
```
