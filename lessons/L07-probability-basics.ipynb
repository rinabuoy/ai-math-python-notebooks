## 1. Sample space and events (លំហសំណាក / ព្រឹត្តិការណ៍)

```python
from itertools import product
S=list(product("HT",repeat=2))
print(S)
A=[s for s in S if s.count("H")==1]      # exactly one head
print("P(A) =", len(A)/len(S))
```

## 3. Random variable and distribution table
X = number of 3s in three dice rolls (Grade 12 advanced).

```python
from fractions import Fraction
outcomes=list(product(range(1,7),repeat=3))
counts={k:0 for k in range(4)}
for o in outcomes: counts[o.count(3)]+=1
dist={k:Fraction(v,len(outcomes)) for k,v in counts.items()}
print(dist, " sum =", sum(dist.values()))
```

## 4. Simulation agrees

```python
rng=np.random.default_rng(0)
rolls=rng.integers(1,7,size=(200000,3))
sim=np.bincount((rolls==3).sum(axis=1),minlength=4)/len(rolls)
print(np.round(sim,4), [round(float(v),4) for v in dist.values()])
```
