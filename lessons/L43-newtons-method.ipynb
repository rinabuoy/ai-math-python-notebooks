## 1. Root finding: √2

```python
f=lambda x:x**2-2; df=lambda x:2*x
x=1.0
for k in range(6): x=x-f(x)/df(x); print(k+1,x)
```

## 2. Minimising g(x)=e^x−ln x

```python
g1=lambda x:np.exp(x)-1/x; g2=lambda x:np.exp(x)+1/x**2
x=0.05; seq=[x]
for _ in range(7): x=x-g1(x)/g2(x); seq.append(x)
print(np.round(seq,4))     # 0.05 0.097 0.183 0.320 0.477 0.558 0.567
```

## 3. Two variables: f = x⁴+0.8y⁴+4x²+2y²−xy−0.2x²y

```python
def grad(p):
    x,y=p; return np.array([4*x**3+8*x-y-0.4*x*y, 3.2*y**3+4*y-x-0.2*x**2])
def hess(p):
    x,y=p; return np.array([[12*x**2+8-0.4*y,-1-0.4*x],[-1-0.4*x,9.6*y**2+4]])
p=np.array([4.,4.]); print(grad(p),hess(p))   # (213.6, 277.6) ...
for k in range(8): p=p-np.linalg.solve(hess(p),grad(p)); 
print(p)      # -> (0,0) in ~8 steps
```
