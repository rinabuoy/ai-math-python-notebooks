## 1. Product and sum rules

```python
import itertools, math
print(4*3)                                   # 4 shirts x 3 shorts
dice=list(itertools.product(range(1,7),repeat=2))
A={d for d in dice if sum(d)==4}; B={d for d in dice if d[0]*d[1]==4}
print(len(A),len(B),len(A|B),len(A)+len(B)-len(A&B))   # 3 3 5 5
```

## 2. Permutations and combinations

```python
P=lambda n,r: math.factorial(n)//math.factorial(n-r)
C=lambda n,r: P(n,r)//math.factorial(r)
print(P(6,4),P(7,3),C(4,2))
assert P(6,4)==len(list(itertools.permutations(range(6),4)))
assert C(4,2)==len(list(itertools.combinations(range(4),2)))
assert C(10,3)==math.comb(10,3)
```

## 3. Pascal's triangle and the binomial theorem

```python
rows=[[math.comb(n,r) for r in range(n+1)] for n in range(7)]
for r in rows: print(r)
assert all(rows[n][r]==rows[n-1][r-1]+rows[n-1][r] for n in range(2,7) for r in range(1,n))
a=sp.symbols('a'); print(sp.expand((1+a)**4))
assert all(sum(row)==2**n for n,row in enumerate(rows))
```

## 4. Counting → probability

```python
n=23
p_no=np.prod([(365-k)/365 for k in range(n)]); print(1-p_no)       # 0.507
print(P(365,n)/365**n==p_no or abs(P(365,n)/365**n-p_no)<1e-12)
print(math.comb(10,3)*0.5**10)    # P(exactly 3 heads in 10 fair tosses)
```

## 5. Circular arrangements

```python
def circ(n): return math.factorial(n-1)
print(circ(6),2*circ(5))   # 120 and the 2 adjacent women: 48 = 2! * 4!
seats=set()
for perm in itertools.permutations(range(5)):
    k=perm.index(0); seats.add(perm[k:]+perm[:k])
print(len(seats),circ(5))
```

**Textbook example (G10 v2 p. 85):** if $(a-\tfrac12x)^6=b-96x+cx^2+\dots$ find $a,b,c$ by matching binomial coefficients.

```python
a,x=sp.symbols('a x'); co=sp.Poly(sp.expand((a-x/2)**6),x).all_coeffs()[::-1]
sol=[s for s in sp.solve(sp.Eq(co[1],-96),a) if s.is_real][0]
print(sol,co[0].subs(a,sol),co[2].subs(a,sol)); assert (sol,co[0].subs(a,sol),co[2].subs(a,sol))==(2,64,60)
```
