## 1. y = sin(u), u = x² − 3

```python
x=sp.symbols('x'); u=x**2-3; y=sp.sin(u)
print(sp.diff(y,x), "=", 2*x*sp.cos(u))
```

## 2. Numerical check with three layers: f(g(h(t)))

```python
h=lambda t:t**2; g=lambda t:np.exp(t); f=lambda t:np.log(1+t)
F=lambda t:f(g(h(t)))
t0=0.6; analytic=(1/(1+g(h(t0))))*g(h(t0))*(2*t0)
print((F(t0+1e-6)-F(t0-1e-6))/2e-6, analytic)
```

## 3. Temperature story: dT/dt = dT/dh · dh/dt

```python
dTdh=-6.5; dhdt=0.4     # degrees per km, km per hour
print("dT/dt =", dTdh*dhdt, "degrees per hour")
```
