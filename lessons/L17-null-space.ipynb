## 1. Solutions of A x = 0

```python
from scipy.linalg import null_space
A=np.array([[1,1],[2,2]],float)
N=null_space(A); print(N.T, "A@N ~ 0:", np.allclose(A@N,0))   # direction (1,-1)/sqrt2
```

## 2. Dimensions 0, 1, 2 for rank 2, 1, 0

```python
for A in ([[3,1],[1,2]],[[1,1],[2,2]],[[0,0],[0,0]]):
    print(A, "nullity", null_space(np.array(A,float)).shape[1])
```

## 3. Picture: null space is squashed to the origin

```python
A=np.array([[1,1],[2,2]],float); t=np.linspace(-2,2,5)
line=np.outer(null_space(A)[:,0],t)
print(np.round(A@line,10))   # all zeros
plt.plot(line[0],line[1],'o-'); plt.axis('equal'); plt.title("null space = line x+y=0"); plt.show()
```
