## 2. Loop vs vectorised speed

```python
import time
x=np.random.default_rng(0).normal(size=1_000_000); y=np.random.default_rng(1).normal(size=1_000_000)
t=time.time(); s=sum(a*b for a,b in zip(x,y)); t1=time.time()-t
t=time.time(); s2=x@y; t2=time.time()-t
print(np.isclose(s,s2), f"loop {t1:.3f}s vs vectorised {t2:.5f}s")
```

## 3. Geometry: u·v = |u||v|cosθ

```python
def angle(u,v):
    u,v=np.array(u,float),np.array(v,float)
    return np.degrees(np.arccos(u@v/(np.linalg.norm(u)*np.linalg.norm(v))))
print(angle([6,2],[-1,3]), angle([2,4],[6,2]), angle([-4,1],[6,2]))   # 90, <90, >90
print((np.array([6,2])@np.array([-1,3]))==0)
```

```python
docs=np.array([[1,1,0],[2,2,0],[0,1,3]],float)
unit=docs/np.linalg.norm(docs,axis=1,keepdims=True); print(np.round(unit@unit.T,3))   # cosine similarity matrix
```

## Cambodian textbook corner (កម្មវិធីសិក្សាកម្ពុជា)
Grade 12 advanced proofs use cos θ = u·v /(|u||v|) — the same formula as above.

```python
print(angle([1,-4,1],[2,3,0]))
```

**Grade 11 advanced, ch. 8 §4 (pp. 202–207):** the book *defines* $\vec u\cdot\vec v=|\vec u||\vec v|\cos\theta$ and then *proves* the coordinate form $xx'+yy'+zz'$. Check both agree, plus Cauchy–Schwarz and the triangle-area formula.

```python
u=np.array([-4,0,2.]); v=np.array([2,0,-1.])
print("coordinate form:",u@v,"  geometric:",np.linalg.norm(u)*np.linalg.norm(v)*np.cos(np.pi))   # book: -10, theta=pi
print("textbook pair:",np.array([3,-1,2])@np.array([1,-1,-2]),"-> perpendicular")
rng=np.random.default_rng(1)
for _ in range(1000):
    a,b=rng.normal(size=3),rng.normal(size=3)
    assert abs(a@b)<=np.linalg.norm(a)*np.linalg.norm(b)+1e-12            # Cauchy-Schwarz
    assert np.linalg.norm(a+b)<=np.linalg.norm(a)+np.linalg.norm(b)+1e-12  # triangle inequality
    S=0.5*np.sqrt((a@a)*(b@b)-(a@b)**2)
    assert np.isclose(S,0.5*np.linalg.norm(np.cross(a,b)))                 # area formula = half |a x b|
print("inequalities and area formula verified on 1000 random pairs")
```
