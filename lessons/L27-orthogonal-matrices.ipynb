```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Rotation matrices are orthogonal

```python
a=np.radians(45); Q=np.array([[np.cos(a),-np.sin(a)],[np.sin(a),np.cos(a)]])
print(np.round(Q.T@Q,10), np.linalg.det(Q))
```

## 2. Lengths and dot products are preserved

```python
u,v=np.array([3,4.]),np.array([1,-2.])
print(np.linalg.norm(Q@u),np.linalg.norm(u), (Q@u)@(Q@v), u@v)
```

## 3. Orthogonal columns (the textbook's [[6,-1],[2,3]]) and QR

```python
A=np.array([[6,-1],[2,3]],float); print("columns dot:",A[:,0]@A[:,1])
Qr,R=np.linalg.qr(A); print(np.round(Qr,4)); print(np.allclose(Qr.T@Qr,np.eye(2)))
```
