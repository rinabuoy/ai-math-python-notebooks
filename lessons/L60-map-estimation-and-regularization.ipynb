## 1. MLE for 6 heads, 4 tails

```python
p=np.linspace(0.001,0.999,999); L=p**6*(1-p)**4
print(p[L.argmax()], p[np.log(L).argmax()])      # same maximiser, 0.6
assert abs(p[L.argmax()]-0.6)<0.002
```

## 2. MAP with a Beta(2,2) prior

```python
a=b=2; post=p**(6+a-1)*(1-p)**(4+b-1); print(p[post.argmax()], (6+a-1)/(10+a+b-2))   # 7/12
assert abs(p[post.argmax()]-7/12)<0.002
```

## 3. Ridge regression = MAP with a Gaussian prior

```python
from scipy.optimize import minimize
rng=np.random.default_rng(3); X=rng.normal(size=(40,3)); w_true=np.array([1.,-2.,0.5]); sig=1.0; y=X@w_true+rng.normal(0,sig,40)
tau=0.5; lam=sig**2/tau**2
ridge=np.linalg.solve(X.T@X+lam*np.eye(3),X.T@y)
neg_post=lambda w: ((y-X@w)**2).sum()/(2*sig**2)+(w**2).sum()/(2*tau**2)
mapw=minimize(neg_post,np.zeros(3)).x
print(ridge,mapw); assert np.allclose(ridge,mapw,atol=1e-4)
print(np.linalg.norm(ridge),np.linalg.norm(np.linalg.lstsq(X,y,rcond=None)[0]))   # shrunk
```

## 4. Overfitting and lambda

```python
xs=np.linspace(-1,1,12); ys=xs**2+rng.normal(0,0.1,12); V=np.vander(xs,10)
for lam in (0,1e-3,1e-1):
    w=np.linalg.solve(V.T@V+lam*np.eye(10),V.T@ys); print(lam,np.abs(w).max())
```
