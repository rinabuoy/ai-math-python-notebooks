## 1. A function in Python
A function gives **exactly one** output per input.

```python
def f(x): return x**2 + 2*x
print(f(2), f(0), f(-3))   # textbook values 8, 0, 3
```

## 2. Domain: where does the formula make sense?

```python
def g(x):
    return 1/(x-2) if x != 2 else float('nan')
def h(x):
    return np.sqrt(x) if x >= 0 else float('nan')
print([g(v) for v in (0, 1, 2, 3)])
print([h(v) for v in (-1, 0, 4)])
```

## 3. Is y² = x a function? (it is not)

```python
x = 9
ys = [y for y in np.linspace(-5, 5, 1001) if np.isclose(y**2, x)]
print("outputs for x=9:", np.round(ys, 3))   # two values -> not a function
```

```python
xs = np.linspace(-3, 3, 200)
plt.plot(xs, xs**2 + 2*xs); plt.title("f(x)=x^2+2x"); plt.grid(True); plt.show()
```
