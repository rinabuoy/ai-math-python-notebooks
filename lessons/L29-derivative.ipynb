```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. From average to instantaneous rate

```python
f=lambda x: x**2-1
for h in (1,0.1,0.01,0.001,1e-6): print(h,(f(2+h)-f(2))/h)   # -> 4 (Grade 12 basic)
```

## 2. Symbolic derivative

```python
x=sp.symbols('x')
for expr in (x**2, x**3, 1/x, x**5-7*x**3+x-10, sp.sin(x)): print(expr,"->",sp.diff(expr,x))
```

## 3. Tangent line

```python
xs=np.linspace(0,4,100); x0=2; slope=2*x0
plt.plot(xs,f(xs),label='f'); plt.plot(xs,f(x0)+slope*(xs-x0),'--',label='tangent at 2'); plt.legend(); plt.grid(True); plt.show()
```

## 4. Higher derivatives

```python
g=3*x**5-7*x**3+x-10
print([sp.diff(g,x,k) for k in range(1,7)])   # f^(6) = 0
```

## Cambodian textbook corner (កម្មវិធីសិក្សាកម្ពុជា)
Grade 12 basic p. 44, 53: f(x)=x²−1 ⇒ f'(2)=4; f(x)=3x⁵−7x³+x−10 ⇒ f'=15x⁴−21x²+1 … f⁽⁶⁾=0.

```python
print(sp.diff(g,x), sp.diff(g,x,6))
```
