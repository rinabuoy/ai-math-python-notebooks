```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Implementation

```python
def cramer(A,b):
    A=np.array(A,float); b=np.array(b,float)
    D=np.linalg.det(A)
    if abs(D)<1e-12: return None, "D=0: no unique solution"
    x=[]
    for j in range(len(b)):
        Aj=A.copy(); Aj[:,j]=b; x.append(np.linalg.det(Aj)/D)
    return np.array(x), "ok"
print(cramer([[3,7],[4,-2]],[-6,26]))
print(np.linalg.solve([[3,7],[4,-2]],[-6,26]))
```

## 2. Three unknowns and the D = 0 case

```python
print(cramer([[1,2,3],[-2,1,1],[4,-1,0]],[2,1,3]))
print(cramer([[1,1],[2,2]],[10,24]))
```

## 3. Cost: Cramer's rule vs elimination

```python
import time
A=np.random.default_rng(0).normal(size=(7,7)); b=np.ones(7)
t=time.time(); cramer(A,b); t1=time.time()-t
t=time.time(); np.linalg.solve(A,b); t2=time.time()-t
print(f"cramer {t1:.5f}s  solve {t2:.5f}s  (grows much worse with size)")
```
