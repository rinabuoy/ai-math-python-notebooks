```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Is it a basis?

```python
def is_basis(vectors):
    V=np.array(vectors,float); return V.shape[0]==V.shape[1] and np.linalg.matrix_rank(V)==V.shape[0]
print(is_basis([[3,1],[1,2]]), is_basis([[1,2],[2,4]]))
```

## 2. Coordinates in a basis: solve B c = v

```python
B=np.array([[3,1],[1,2]],float).T   # basis vectors (3,1),(1,2) as columns
v=np.array([4,3]); c=np.linalg.solve(B,v); print(c, B@c)
```

## 3. Span as a picture

```python
rng=np.random.default_rng(0); c=rng.uniform(-1,1,size=(2,200))
pts=B@c; plt.scatter(*pts,s=5); plt.axis('equal'); plt.title("span of (3,1),(1,2) = whole plane region"); plt.show()
print("dimension of span:", np.linalg.matrix_rank(np.array([[1,2],[2,4],[3,6]],float)))
```
