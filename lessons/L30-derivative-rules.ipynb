```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Numerical derivative helper

```python
def nd(f,x,h=1e-6): return (f(x+h)-f(x-h))/(2*h)
x0=0.7
f,g=np.sin,lambda t:t**3
print("sum    ",nd(lambda t:f(t)+g(t),x0), np.cos(x0)+3*x0**2)
print("scalar ",nd(lambda t:5*g(t),x0), 5*3*x0**2)
print("product",nd(lambda t:f(t)*g(t),x0), np.cos(x0)*x0**3+np.sin(x0)*3*x0**2)
print("chain  ",nd(lambda t:np.sin(t**2-3),x0), 2*x0*np.cos(x0**2-3))
```

## 2. SymPy knows them all

```python
x=sp.symbols('x')
exprs=[sp.sin(x**2-3),sp.cos(x)**2+x*sp.sin(x),sp.tan(x),sp.log(x),sp.exp(x)*sp.log(x),sp.log(x+sp.sqrt(1+x**2)),x**2*sp.cos(3*x+1)]
for e in exprs: print(e,"->",sp.simplify(sp.diff(e,x)))
```
