```python
import numpy as np
import sympy as sp
import matplotlib
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
%matplotlib inline
```

## 1. Profit maximisation  P(x)=400(x−2)(15−x)

```python
x=sp.symbols('x'); P=400*(x-2)*(15-x)
xs=sp.solve(sp.diff(P,x),x); print(xs, P.subs(x,xs[0]))   # 8.5 -> 16900
```

## 2. Triangle of perimeter 2p, one side 3p/4: maximise area (p=40)

```python
def area(xv,p=40):
    y=(5*p-4*xv)/4; s=p; return np.sqrt(s*(s-xv)*(s-y)*(s-3*p/4))
xs=np.linspace(10.01,39.99,4000)
best=xs[np.argmax([area(v) for v in xs])]; print(best, 5*40/8)   # 25
```

## 3. Cone of fixed volume: minimise lateral area  → height/radius = √2

```python
V=1.0
def lateral(r): h=3*V/(np.pi*r**2); return np.pi*r*np.sqrt(r**2+h**2)
r=np.linspace(0.3,2,20000); rb=r[np.argmin(lateral(r))]; hb=3*V/(np.pi*rb**2); print(hb/rb, np.sqrt(2))
```

## 4. No closed form: minimise e^x − ln x

```python
from scipy.optimize import minimize_scalar
res=minimize_scalar(lambda t:np.exp(t)-np.log(t),bounds=(0.01,3),method='bounded'); print(res.x)   # 0.5671 (Omega constant)
```

**Grade 11 advanced ch. 5:** use the variation table to count solutions of $f(x)=\lambda$ (here $x^4-2x^2-m=0$).

```python
x=sp.symbols('x',real=True)
for m in (1,0,-0.5,-1,-2): print('m =',m,'->',len(set(sp.real_roots(sp.Poly(x**4-2*x**2-sp.nsimplify(m),x)))),'real roots')
f=x**3+3*x**2-2; print([(c,f.subs(x,c)) for c in sp.solve(sp.diff(f,x),x)])   # local max 2 at -2, local min -2 at 0
assert len(set(sp.real_roots(sp.Poly(x**4-2*x**2,x))))==3
```
