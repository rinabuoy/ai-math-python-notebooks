## 1. First-order recurrence $a_{n+1}=1+4a_n$

```python
a=[1]
for _ in range(7): a.append(1+4*a[-1])
print(a,[(4**i-1)//3 for i in range(1,9)]); assert a==[(4**i-1)//3 for i in range(1,9)]
```

## 2. Second-order recurrence via characteristic roots

```python
x=sp.symbols('x'); print(sp.solve(x**2-3*x+2,x))         # roots 1 and 2 (alpha+beta=3, alpha*beta=2)
s=[1,3]
for _ in range(8): s.append(3*s[-1]-2*s[-2])
print(s); assert s==[2**i-1 for i in range(1,11)]
n=sp.symbols('n',integer=True); f=sp.Function('f'); print(sp.rsolve(f(n+2)-3*f(n+1)+2*f(n),f(n),{f(1):1,f(2):3}))
```

## 3. Checking an induction claim numerically (then it still needs a proof!)

```python
print(all(sum(2*i-1 for i in range(1,m+1))==m**2 for m in range(1,500)))
print(all((4**m+2)%3==0 for m in range(1,500)), all((1+0.3)**m>=1+m*.3+m*(m-1)/2*.09 for m in range(1,50)))
```

## 4. Gradient descent as a recurrence

```python
def gd(alpha,x0=1.,steps=20):
    xs=[x0]
    for _ in range(steps): xs.append(xs[-1]-alpha*xs[-1])      # f(x)=x^2/2, f'(x)=x
    return np.array(xs)
for al in (0.1,1.0,1.9,2.1): xs=gd(al); print(al,xs[-1],np.isclose(xs[-1],(1-al)**20))
```
