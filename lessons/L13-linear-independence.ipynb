## 1. Row1 + Row2 = Row3

```python
M=np.array([[1,0,0],[0,1,0],[1,1,0]])
print(np.linalg.matrix_rank(M), "independent rows out of", len(M))
```

## 2. Finding the relation: solve c1*r1 + c2*r2 = r3

```python
rows=np.array([[1,0,1],[0,1,0],[3,2,3]],float)   # from the Coursera quiz
c,res,rk,sv=np.linalg.lstsq(rows[:2].T,rows[2],rcond=None); print(c, "residual", res)  # 3 R1 + 2 R2 = R3
```

## 3. Independent example

```python
I=np.array([[1,1,1],[1,2,1],[1,1,2]]); print(np.linalg.matrix_rank(I), np.linalg.det(I))
```

```python
def independent(vectors): 
    return np.linalg.matrix_rank(np.array(vectors,float))==len(vectors)
print(independent([[1,0],[0,1]]), independent([[1,2],[2,4]]))
```
